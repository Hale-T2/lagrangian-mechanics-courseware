## 1. Philosophical & Mathematical Comparison

| Feature | Newtonian Mechanics | Lagrangian Mechanics |
| :--- | :--- | :--- |
| **Fundamental Quantity** | Vector Forces $\mathbf{F}$ | Scalar Energy $L = T - V$ |
| **Governing Equation** | $\mathbf{F}_{net} = m\mathbf{a}$ | $\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i}\right) - \frac{\partial L}{\partial q_i} = 0$ |
| **Perspective** | Differential (instantaneous local cause & effect) | Integral (global path optimization / stationary action) |
| **Coordinates** | Fixed 3D Cartesian coordinates $(x, y, z)$ | Arbitrary Generalized Coordinates ($q_i$, e.g., angles $\theta$) |
| **Constraint Forces** | Must be explicitly calculated (Normal force $N$, Tension $T$) | Automatically eliminated from equations of motion |
| **Curvilinear Geometry** | Complex transformation of vector components | Trivial substitution into kinetic energy $T$ |

## 2. Deriving the Euler-Lagrange Equation with SymPy

The action $S$ is defined as the integral of the Lagrangian $L(q, \dot{q}, t)$ over time:

$$S = \int_{t_1}^{t_2} L(q(t), \dot{q}(t), t) \, dt$$

To find the path $q(t)$ that makes $S$ stationary, we consider a small variation $q(t) \to q(t) + \epsilon \eta(t)$ where $\eta(t_1) = \eta(t_2) = 0$. Setting $\frac{dS}{d\epsilon}\Big|_{\epsilon=0} = 0$ yields the celebrated **Euler-Lagrange equation**:

$$\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}}\right) - \frac{\partial L}{\partial q} = 0$$


```python
# Symbolic derivation of Euler-Lagrange for a generic 1D particle
t = sp.Symbol('t', real=True)
m, k = sp.symbols('m k', positive=True)
x = sp.Function('x')(t)
v = sp.diff(x, t)

# Define Kinetic (T) and Potential (V) Energy
T = sp.Rational(1, 2) * m * v**2
V = sp.Rational(1, 2) * k * x**2
L = T - V

print('Lagrangian L = T - V:')
sp.pprint(L)

# Euler-Lagrange terms
dL_dv = sp.diff(L, v)           # Partial derivative wrt velocity (Canonical Momentum)
dt_dL_dv = sp.diff(dL_dv, t)    # Total time derivative
dL_dx = sp.diff(L, x)           # Partial derivative wrt position (Generalized Force)

eq_of_motion = sp.Eq(dt_dL_dv - dL_dx, 0)
print('\nDerived Equation of Motion:')
sp.pprint(eq_of_motion)
```
